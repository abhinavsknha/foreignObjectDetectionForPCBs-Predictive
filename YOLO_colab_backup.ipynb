```python
import os
import shutil
import random
from pathlib import Path

# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = Path("/content/drive/MyDrive/DataSet_CourseProject")

ORIGINAL_DIR = BASE_DIR / "Classified_Dataset"
NEW_DIR = BASE_DIR / "PCB_YOLO_Dataset"

TRAIN_IMG_DIR = ORIGINAL_DIR / "images" / "train"
TRAIN_LBL_DIR = ORIGINAL_DIR / "labels" / "train"

VAL_IMG_DIR = ORIGINAL_DIR / "images" / "val"
VAL_LBL_DIR = ORIGINAL_DIR / "labels" / "val"

# ============================================================
# 2. SETTINGS
# ============================================================

TRAIN_RATIO = 0.85
RANDOM_SEED = 42

random.seed(RANDOM_SEED)

# ============================================================
# 3. CREATE NEW DATASET DIRECTORIES
# ============================================================

for split in ["train", "val", "test"]:
    (NEW_DIR / "images" / split).mkdir(parents=True, exist_ok=True)
    (NEW_DIR / "labels" / split).mkdir(parents=True, exist_ok=True)

print("New dataset directory created:")
print(NEW_DIR)

# ============================================================
# 4. SPLIT ORIGINAL TRAINING DATA CLASS-WISE
# ============================================================

classes = sorted([
    d.name for d in TRAIN_IMG_DIR.iterdir()
    if d.is_dir() and d.name.isdigit()
])

print("\nClasses found:", classes)

total_train = 0
total_test = 0

for class_id in classes:

    class_img_dir = TRAIN_IMG_DIR / class_id
    class_lbl_dir = TRAIN_LBL_DIR / class_id

    # Get images
    images = [
        f for f in class_img_dir.iterdir()
        if f.is_file() and f.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
    ]

    random.shuffle(images)

    split_index = int(len(images) * TRAIN_RATIO)

    train_images = images[:split_index]
    test_images = images[split_index:]

    # Create destination class folders
    new_train_img_dir = NEW_DIR / "images" / "train" / class_id
    new_train_lbl_dir = NEW_DIR / "labels" / "train" / class_id

    new_test_img_dir = NEW_DIR / "images" / "test" / class_id
    new_test_lbl_dir = NEW_DIR / "labels" / "test" / class_id

    for directory in [
        new_train_img_dir,
        new_train_lbl_dir,
        new_test_img_dir,
        new_test_lbl_dir
    ]:
        directory.mkdir(parents=True, exist_ok=True)

    # -------------------------
    # Copy TRAIN
    # -------------------------

    for img_path in train_images:

        label_path = class_lbl_dir / f"{img_path.stem}.txt"

        shutil.copy2(
            img_path,
            new_train_img_dir / img_path.name
        )

        if label_path.exists():
            shutil.copy2(
                label_path,
                new_train_lbl_dir / label_path.name
            )

    # -------------------------
    # Copy TEST
    # -------------------------

    for img_path in test_images:

        label_path = class_lbl_dir / f"{img_path.stem}.txt"

        shutil.copy2(
            img_path,
            new_test_img_dir / img_path.name
        )

        if label_path.exists():
            shutil.copy2(
                label_path,
                new_test_lbl_dir / label_path.name
            )

    total_train += len(train_images)
    total_test += len(test_images)

    print(
        f"Class {class_id}: "
        f"{len(images)} total → "
        f"{len(train_images)} train / "
        f"{len(test_images)} test"
    )

# ============================================================
# 5. COPY EXISTING VALIDATION DATA UNCHANGED
# ============================================================

print("\nCopying existing validation dataset...")

for class_id in classes:

    src_img_dir = VAL_IMG_DIR / class_id
    src_lbl_dir = VAL_LBL_DIR / class_id

    dst_img_dir = NEW_DIR / "images" / "val" / class_id
    dst_lbl_dir = NEW_DIR / "labels" / "val" / class_id

    dst_img_dir.mkdir(parents=True, exist_ok=True)
    dst_lbl_dir.mkdir(parents=True, exist_ok=True)

    # Images
    if src_img_dir.exists():
        for img_path in src_img_dir.iterdir():
            if img_path.is_file():
                shutil.copy2(
                    img_path,
                    dst_img_dir / img_path.name
                )

    # Labels
    if src_lbl_dir.exists():
        for label_path in src_lbl_dir.iterdir():
            if label_path.is_file():
                shutil.copy2(
                    label_path,
                    dst_lbl_dir / label_path.name
                )

print("Validation copied successfully.")

# ============================================================
# 6. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("DATASET SPLIT COMPLETE")
print("=" * 60)

print(f"Original dataset : {ORIGINAL_DIR}")
print(f"New dataset      : {NEW_DIR}")

print(f"\nNew training images : {total_train}")
print(f"New testing images  : {total_test}")

print("\nDataset structure:")

print("""
PCB_YOLO_Dataset/
│
├── images/
│   ├── train/
│   ├── val/
│   └── test/
│
└── labels/
    ├── train/
    ├── val/
    └── test/
""")

print("Original dataset has NOT been modified.")
```

```python
import os
import shutil
import random
from pathlib import Path

# ============================================================
# PATHS
# ============================================================

BASE_DIR = Path("/content/drive/MyDrive/DataSet_CourseProject")

ORIGINAL_DIR = BASE_DIR / "Classified_Dataset"
NEW_DIR = BASE_DIR / "PCB_YOLO_Dataset"

IMAGE_TRAIN_DIR = ORIGINAL_DIR / "images" / "train"
LABEL_TRAIN_DIR = ORIGINAL_DIR / "labels" / "train"

IMAGE_VAL_DIR = ORIGINAL_DIR / "images" / "val"
LABEL_VAL_DIR = ORIGINAL_DIR / "labels" / "val"

# ============================================================
# SETTINGS
# ============================================================

TRAIN_RATIO = 0.85
RANDOM_SEED = 42

random.seed(RANDOM_SEED)

# ============================================================
# DELETE PREVIOUS INCOMPLETE WORKING DATASET
# ============================================================

if NEW_DIR.exists():
    print("Removing previous incomplete PCB_YOLO_Dataset...")
    shutil.rmtree(NEW_DIR)

# ============================================================
# CREATE DIRECTORY STRUCTURE
# ============================================================

for split in ["train", "val", "test"]:
    (NEW_DIR / "images" / split).mkdir(parents=True, exist_ok=True)
    (NEW_DIR / "labels" / split).mkdir(parents=True, exist_ok=True)

print("Created:", NEW_DIR)

# ============================================================
# IMAGE FOLDER → YOLO CLASS MAPPING
# ============================================================

CLASS_MAPPING = {
    "1 SH (Solder Pads)": "1",
    "2 SP (Shorts)": "0",
    "3 SC (Scratches)": "2",
    "4 OP (Opening)": "3",
    "5 MB (Mask Break)": "4",
    "6 HB (Hole Break)": "5",
    "7 CS (Cremation)": "6",
    "8 CFO (Component Foreign Object)": "7",
    "9 BMFO (Bare Metal-Free Offset)": "8"
}

# ============================================================
# 1. SPLIT ORIGINAL TRAINING DATA
# ============================================================

print("\n" + "=" * 70)
print("CREATING 85% TRAIN / 15% TEST SPLIT")
print("=" * 70)

total_train = 0
total_test = 0

for image_folder_name, class_id in CLASS_MAPPING.items():

    source_image_dir = IMAGE_TRAIN_DIR / image_folder_name
    source_label_dir = LABEL_TRAIN_DIR / class_id

    if not source_image_dir.exists():
        print(f"\nWARNING: Missing image folder: {image_folder_name}")
        continue

    images = [
        f for f in source_image_dir.iterdir()
        if f.is_file()
        and f.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
    ]

    # Reproducible shuffle
    random.shuffle(images)

    split_index = int(len(images) * TRAIN_RATIO)

    train_images = images[:split_index]
    test_images = images[split_index:]

    # Destination directories
    train_img_dir = NEW_DIR / "images" / "train" / class_id
    train_lbl_dir = NEW_DIR / "labels" / "train" / class_id

    test_img_dir = NEW_DIR / "images" / "test" / class_id
    test_lbl_dir = NEW_DIR / "labels" / "test" / class_id

    train_img_dir.mkdir(parents=True, exist_ok=True)
    train_lbl_dir.mkdir(parents=True, exist_ok=True)

    test_img_dir.mkdir(parents=True, exist_ok=True)
    test_lbl_dir.mkdir(parents=True, exist_ok=True)

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    missing_train_labels = 0

    for image_path in train_images:

        label_path = source_label_dir / f"{image_path.stem}.txt"

        shutil.copy2(
            image_path,
            train_img_dir / image_path.name
        )

        if label_path.exists():
            shutil.copy2(
                label_path,
                train_lbl_dir / label_path.name
            )
        else:
            missing_train_labels += 1

    # --------------------------------------------------------
    # TEST
    # --------------------------------------------------------

    missing_test_labels = 0

    for image_path in test_images:

        label_path = source_label_dir / f"{image_path.stem}.txt"

        shutil.copy2(
            image_path,
            test_img_dir / image_path.name
        )

        if label_path.exists():
            shutil.copy2(
                label_path,
                test_lbl_dir / label_path.name
            )
        else:
            missing_test_labels += 1

    total_train += len(train_images)
    total_test += len(test_images)

    print(
        f"{class_id} | {image_folder_name:<40} "
        f"| Total: {len(images):4d} "
        f"| Train: {len(train_images):4d} "
        f"| Test: {len(test_images):4d}"
    )

    if missing_train_labels or missing_test_labels:
        print(
            f"     WARNING: Missing labels → "
            f"train={missing_train_labels}, test={missing_test_labels}"
        )

# ============================================================
# 2. COPY EXISTING VALIDATION DATA
# ============================================================

print("\n" + "=" * 70)
print("COPYING EXISTING VALIDATION DATA")
print("=" * 70)

val_images_total = 0
val_labels_total = 0

for image_folder_name, class_id in CLASS_MAPPING.items():

    source_image_dir = IMAGE_VAL_DIR / image_folder_name
    source_label_dir = LABEL_VAL_DIR / class_id

    destination_image_dir = NEW_DIR / "images" / "val" / class_id
    destination_label_dir = NEW_DIR / "labels" / "val" / class_id

    destination_image_dir.mkdir(parents=True, exist_ok=True)
    destination_label_dir.mkdir(parents=True, exist_ok=True)

    # Copy images
    if source_image_dir.exists():

        for image_path in source_image_dir.iterdir():

            if image_path.is_file():
                shutil.copy2(
                    image_path,
                    destination_image_dir / image_path.name
                )
                val_images_total += 1

    # Copy labels
    if source_label_dir.exists():

        for label_path in source_label_dir.iterdir():

            if label_path.is_file():
                shutil.copy2(
                    label_path,
                    destination_label_dir / label_path.name
                )
                val_labels_total += 1

print(f"Validation images copied: {val_images_total}")
print(f"Validation labels copied : {val_labels_total}")

# ============================================================
# 3. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("DATASET CREATION COMPLETE")
print("=" * 70)

print(f"""
Original dataset:
{ORIGINAL_DIR}

New working dataset:
{NEW_DIR}

Training images : {total_train}
Testing images  : {total_test}
Validation imgs : {val_images_total}

Split:
85% → TRAIN
15% → TEST
100% existing VAL → VALIDATION

Original dataset has NOT been modified.
""")
```

```python
from pathlib import Path
from collections import defaultdict

BASE_DIR = Path("/content/drive/MyDrive/DataSet_CourseProject")
DATASET = BASE_DIR / "Classified_Dataset"

IMAGE_TRAIN = DATASET / "images" / "train"
LABEL_TRAIN = DATASET / "labels" / "train"

# ------------------------------------------------------------
# Build index of ALL label files
# ------------------------------------------------------------

label_locations = defaultdict(list)

for label_folder in sorted([x for x in LABEL_TRAIN.iterdir() if x.is_dir()]):

    for label_file in label_folder.iterdir():

        if label_file.is_file() and label_file.suffix.lower() == ".txt":
            label_locations[label_file.stem].append(label_folder.name)

# ------------------------------------------------------------
# Check every training image
# ------------------------------------------------------------

print("=" * 75)
print("TRAIN IMAGE ↔ LABEL INTEGRITY CHECK")
print("=" * 75)

total_images = 0
matched_images = 0
missing_images = []
multiple_label_locations = []

for image_folder in sorted([x for x in IMAGE_TRAIN.iterdir() if x.is_dir()]):

    images = [
        x for x in image_folder.iterdir()
        if x.is_file()
        and x.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
    ]

    for image in images:

        total_images += 1

        locations = label_locations.get(image.stem, [])

        if len(locations) == 0:
            missing_images.append(
                (image_folder.name, image.name)
            )

        elif len(locations) == 1:
            matched_images += 1

        else:
            multiple_label_locations.append(
                (image.name, locations)
            )

print(f"\nTotal training images : {total_images}")
print(f"Images with labels    : {matched_images}")
print(f"Missing labels        : {len(missing_images)}")
print(f"Multiple label copies : {len(multiple_label_locations)}")

# ------------------------------------------------------------
# Show missing labels
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("MISSING LABELS")
print("=" * 75)

if missing_images:

    for folder, filename in missing_images[:100]:
        print(f"{folder} → {filename}")

else:
    print("None")

# ------------------------------------------------------------
# Find orphan labels
# ------------------------------------------------------------

all_image_stems = set()

for image_folder in IMAGE_TRAIN.iterdir():

    if not image_folder.is_dir():
        continue

    for image in image_folder.iterdir():

        if image.is_file() and image.suffix.lower() in [
            ".jpg", ".jpeg", ".png", ".bmp"
        ]:
            all_image_stems.add(image.stem)

orphan_labels = []

for label_folder in LABEL_TRAIN.iterdir():

    if not label_folder.is_dir():
        continue

    for label in label_folder.iterdir():

        if label.is_file() and label.suffix.lower() == ".txt":

            if label.stem not in all_image_stems:
                orphan_labels.append(
                    (label_folder.name, label.name)
                )

print("\n" + "=" * 75)
print("ORPHAN LABELS")
print("=" * 75)

print(f"Total orphan labels: {len(orphan_labels)}")

for folder, filename in orphan_labels:
    print(f"{folder} → {filename}")

# ------------------------------------------------------------
# Check multiple label locations
# ------------------------------------------------------------

if multiple_label_locations:

    print("\n" + "=" * 75)
    print("LABELS FOUND IN MULTIPLE FOLDERS")
    print("=" * 75)

    for filename, locations in multiple_label_locations[:50]:
        print(filename, "→", locations)
```

```python
import shutil
import random
from pathlib import Path
from collections import defaultdict

# ============================================================
# PATHS
# ============================================================

BASE_DIR = Path("/content/drive/MyDrive/DataSet_CourseProject")

ORIGINAL_DIR = BASE_DIR / "Classified_Dataset"
NEW_DIR = BASE_DIR / "PCB_YOLO_Dataset"

IMAGE_TRAIN_DIR = ORIGINAL_DIR / "images" / "train"
LABEL_TRAIN_DIR = ORIGINAL_DIR / "labels" / "train"

IMAGE_VAL_DIR = ORIGINAL_DIR / "images" / "val"
LABEL_VAL_DIR = ORIGINAL_DIR / "labels" / "val"

# ============================================================
# SETTINGS
# ============================================================

TRAIN_RATIO = 0.85
RANDOM_SEED = 42

random.seed(RANDOM_SEED)

# ============================================================
# ACTUAL DATASET MAPPING
# ============================================================

CLASS_MAPPING = {
    "1 SH (Solder Pads)": "0",
    "2 SP (Shorts)": "1",
    "3 SC (Scratches)": "2",
    "4 OP (Opening)": "3",
    "5 MB (Mask Break)": "4",
    "6 HB (Hole Break)": "5",
    "7 CS (Cremation)": "6",
    "8 CFO (Component Foreign Object)": "7",
    "9 BMFO (Bare Metal-Free Offset)": "8"
}

# ============================================================
# REMOVE PREVIOUS WORKING DATASET
# ============================================================

if NEW_DIR.exists():
    print("Removing previous incomplete working dataset...")
    shutil.rmtree(NEW_DIR)

# ============================================================
# CREATE NEW STRUCTURE
# ============================================================

for split in ["train", "val", "test"]:
    for data_type in ["images", "labels"]:
        (NEW_DIR / data_type / split).mkdir(
            parents=True,
            exist_ok=True
        )

print("Created new dataset:")
print(NEW_DIR)

# ============================================================
# BUILD LABEL INDEX
# ============================================================
# We support:
#
# image: ABC.jpg
# label: ABC.txt
#
# image: ABC (1).jpg
# label: ABC.txt
#
# image: ABC.jpg
# label: ABC (1).txt
#
# Exact filename matches are always preferred.
# ============================================================

label_files = []

for label_folder in LABEL_TRAIN_DIR.iterdir():

    if not label_folder.is_dir():
        continue

    for label in label_folder.iterdir():

        if label.is_file() and label.suffix.lower() == ".txt":
            label_files.append(label)

# Exact lookup
exact_label_lookup = defaultdict(list)

for label in label_files:
    exact_label_lookup[label.stem].append(label)

# Base-name lookup
# Example:
# "Y_002589 (1)" → "Y_002589"
base_label_lookup = defaultdict(list)

for label in label_files:

    stem = label.stem

    if stem.endswith(" (1)"):
        base_stem = stem[:-4]
    else:
        base_stem = stem

    base_label_lookup[base_stem].append(label)

# ============================================================
# FUNCTION TO FIND MATCHING LABEL
# ============================================================

def find_label(image_path):

    stem = image_path.stem

    # --------------------------------------------------------
    # 1. Exact match
    # --------------------------------------------------------

    exact_matches = exact_label_lookup.get(stem, [])

    if len(exact_matches) == 1:
        return exact_matches[0]

    if len(exact_matches) > 1:
        return None

    # --------------------------------------------------------
    # 2. Image has "(1)" but label doesn't
    # --------------------------------------------------------

    if stem.endswith(" (1)"):

        base_stem = stem[:-4]

        matches = exact_label_lookup.get(base_stem, [])

        if len(matches) == 1:
            return matches[0]

    # --------------------------------------------------------
    # 3. Label has "(1)" but image doesn't
    # --------------------------------------------------------

    matches = base_label_lookup.get(stem, [])

    if len(matches) == 1:
        return matches[0]

    # --------------------------------------------------------
    # No unique match
    # --------------------------------------------------------

    return None

# ============================================================
# VERIFY ALL TRAINING IMAGES FIRST
# ============================================================

print("\n" + "=" * 75)
print("VERIFYING TRAINING IMAGE → LABEL PAIRS")
print("=" * 75)

all_train_pairs = []

missing_labels = []
duplicate_matches = []

for image_folder_name, class_id in CLASS_MAPPING.items():

    source_image_dir = IMAGE_TRAIN_DIR / image_folder_name

    images = [
        x for x in source_image_dir.iterdir()
        if x.is_file()
        and x.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
    ]

    for image in images:

        matches = []

        label = find_label(image)

        if label is None:
            missing_labels.append(
                (image_folder_name, image.name)
            )
        else:
            all_train_pairs.append(
                (image_folder_name, class_id, image, label)
            )

print(f"Total training images : {sum(1 for _ in all_train_pairs) + len(missing_labels)}")
print(f"Matched image-labels   : {len(all_train_pairs)}")
print(f"Missing labels         : {len(missing_labels)}")

if missing_labels:

    print("\nMISSING LABELS:")
    for folder, image in missing_labels:
        print(f"  {folder} → {image}")

    raise RuntimeError(
        "Training dataset contains images without labels. "
        "Dataset creation stopped."
    )

print("\nAll training images have matching labels.")

# ============================================================
# SPLIT TRAINING DATA CLASS-WISE
# ============================================================

print("\n" + "=" * 75)
print("CREATING 85% TRAIN / 15% TEST SPLIT")
print("=" * 75)

total_train = 0
total_test = 0

for image_folder_name, class_id in CLASS_MAPPING.items():

    # Get all pairs belonging to this image class
    class_pairs = [
        pair
        for pair in all_train_pairs
        if pair[0] == image_folder_name
    ]

    random.shuffle(class_pairs)

    split_index = int(len(class_pairs) * TRAIN_RATIO)

    train_pairs = class_pairs[:split_index]
    test_pairs = class_pairs[split_index:]

    # Destination folders
    train_img_dir = NEW_DIR / "images" / "train" / class_id
    train_lbl_dir = NEW_DIR / "labels" / "train" / class_id

    test_img_dir = NEW_DIR / "images" / "test" / class_id
    test_lbl_dir = NEW_DIR / "labels" / "test" / class_id

    train_img_dir.mkdir(parents=True, exist_ok=True)
    train_lbl_dir.mkdir(parents=True, exist_ok=True)

    test_img_dir.mkdir(parents=True, exist_ok=True)
    test_lbl_dir.mkdir(parents=True, exist_ok=True)

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    for _, _, image, label in train_pairs:

        shutil.copy2(
            image,
            train_img_dir / image.name
        )

        shutil.copy2(
            label,
            train_lbl_dir / label.name
        )

    # --------------------------------------------------------
    # TEST
    # --------------------------------------------------------

    for _, _, image, label in test_pairs:

        shutil.copy2(
            image,
            test_img_dir / image.name
        )

        shutil.copy2(
            label,
            test_lbl_dir / label.name
        )

    total_train += len(train_pairs)
    total_test += len(test_pairs)

    print(
        f"Class {class_id} | "
        f"{image_folder_name:<40} | "
        f"Total: {len(class_pairs):4d} | "
        f"Train: {len(train_pairs):4d} | "
        f"Test: {len(test_pairs):4d}"
    )

# ============================================================
# COPY EXISTING VALIDATION DATA
# ============================================================

print("\n" + "=" * 75)
print("COPYING EXISTING VALIDATION DATA")
print("=" * 75)

total_val_images = 0
total_val_labels = 0

# Validation uses numeric folders 0-8
for class_id in CLASS_MAPPING.values():

    source_img_dir = IMAGE_VAL_DIR / class_id
    source_lbl_dir = LABEL_VAL_DIR / class_id

    destination_img_dir = NEW_DIR / "images" / "val" / class_id
    destination_lbl_dir = NEW_DIR / "labels" / "val" / class_id

    destination_img_dir.mkdir(parents=True, exist_ok=True)
    destination_lbl_dir.mkdir(parents=True, exist_ok=True)

    # Images
    if source_img_dir.exists():

        for image in source_img_dir.iterdir():

            if image.is_file():
                shutil.copy2(
                    image,
                    destination_img_dir / image.name
                )
                total_val_images += 1

    # Labels
    if source_lbl_dir.exists():

        for label in source_lbl_dir.iterdir():

            if label.is_file() and label.suffix.lower() == ".txt":
                shutil.copy2(
                    label,
                    destination_lbl_dir / label.name
                )
                total_val_labels += 1

    print(
        f"Class {class_id}: "
        f"{len(list(destination_img_dir.glob('*')))} images | "
        f"{len(list(destination_lbl_dir.glob('*.txt')))} labels"
    )

# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 75)
print("DATASET CREATION COMPLETE")
print("=" * 75)

print(f"""
Original dataset:
{ORIGINAL_DIR}

New YOLO dataset:
{NEW_DIR}

Training images : {total_train}
Testing images  : {total_test}
Validation imgs : {total_val_images}

Validation labels: {total_val_labels}

Split:
85% → TRAIN
15% → TEST
Existing VAL → VALIDATION

Random seed: {RANDOM_SEED}

Original dataset has NOT been modified.
""")

# ============================================================
# BASIC SANITY CHECK
# ============================================================

print("=" * 75)
print("FINAL SANITY CHECK")
print("=" * 75)

for split in ["train", "val", "test"]:

    image_count = sum(
        1
        for f in (NEW_DIR / "images" / split).rglob("*")
        if f.is_file()
        and f.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
    )

    label_count = sum(
        1
        for f in (NEW_DIR / "labels" / split).rglob("*.txt")
        if f.is_file()
    )

    print(
        f"{split.upper():5} → "
        f"{image_count} images | "
        f"{label_count} labels"
    )
```

```python
from pathlib import Path
import shutil

BASE = Path("/content/drive/MyDrive/DataSet_CourseProject")

ORIGINAL_VAL_IMAGES = BASE / "Classified_Dataset" / "images" / "val"
ORIGINAL_VAL_LABELS = BASE / "Classified_Dataset" / "labels" / "val"

NEW_VAL_IMAGES = BASE / "PCB_YOLO_Dataset" / "images" / "val"
NEW_VAL_LABELS = BASE / "PCB_YOLO_Dataset" / "labels" / "val"

# Clear the currently empty validation folders
if NEW_VAL_IMAGES.exists():
    shutil.rmtree(NEW_VAL_IMAGES)

if NEW_VAL_LABELS.exists():
    shutil.rmtree(NEW_VAL_LABELS)

NEW_VAL_IMAGES.mkdir(parents=True, exist_ok=True)
NEW_VAL_LABELS.mkdir(parents=True, exist_ok=True)

# Copy validation class folders
for class_id in [str(i) for i in range(9)]:

    src_img = ORIGINAL_VAL_IMAGES / class_id
    src_lbl = ORIGINAL_VAL_LABELS / class_id

    dst_img = NEW_VAL_IMAGES / class_id
    dst_lbl = NEW_VAL_LABELS / class_id

    dst_img.mkdir(parents=True, exist_ok=True)
    dst_lbl.mkdir(parents=True, exist_ok=True)

    if src_img.exists():
        for file in src_img.iterdir():
            if file.is_file():
                shutil.copy2(file, dst_img / file.name)

    if src_lbl.exists():
        for file in src_lbl.iterdir():
            if file.is_file() and file.suffix.lower() == ".txt":
                shutil.copy2(file, dst_lbl / file.name)

print("Validation dataset restored.")
```

```python
from pathlib import Path
import shutil

BASE = Path("/content/drive/MyDrive/DataSet_CourseProject")

ORIGINAL = BASE / "Classified_Dataset"
NEW = BASE / "PCB_YOLO_Dataset"

for class_id in map(str, range(2, 9)):

    src_img = ORIGINAL / "images" / "val" / class_id
    src_lbl = ORIGINAL / "labels" / "val" / class_id

    dst_img = NEW / "images" / "val" / class_id
    dst_lbl = NEW / "labels" / "val" / class_id

    dst_img.mkdir(parents=True, exist_ok=True)
    dst_lbl.mkdir(parents=True, exist_ok=True)

    print(f"Copying validation class {class_id}...")

    # Images
    for file in src_img.iterdir():
        if file.is_file():
            shutil.copy2(file, dst_img / file.name)

    # Labels
    for file in src_lbl.iterdir():
        if file.is_file() and file.suffix.lower() == ".txt":
            shutil.copy2(file, dst_lbl / file.name)

    print(
        f"  Images: {len(list(dst_img.iterdir()))}"
        f" | Labels: {len(list(dst_lbl.glob('*.txt')))}"
    )

print("\nRemaining validation classes copied.")
```

```python
from pathlib import Path

BASE = Path("/content/drive/MyDrive/DataSet_CourseProject")

for dataset_name in ["Classified_Dataset", "PCB_YOLO_Dataset"]:

    dataset = BASE / dataset_name

    print("\n" + "=" * 70)
    print(dataset_name)
    print("=" * 70)

    for split in ["train", "val", "test"]:

        image_root = dataset / "images" / split

        if not image_root.exists():
            print(f"\n{split.upper()}: folder missing")
            continue

        print(f"\n{split.upper()}")

        total = 0

        for class_dir in sorted(
            [x for x in image_root.iterdir() if x.is_dir()],
            key=lambda x: int(x.name) if x.name.isdigit() else 999
        ):

            count = sum(
                1 for f in class_dir.iterdir()
                if f.is_file()
                and f.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
            )

            total += count
            print(f"  Class {class_dir.name}: {count}")

        print(f"  TOTAL: {total}")
```

```python
from ultralytics import YOLO
import pandas as pd

# Load your trained BEST model
model = YOLO(
    "/content/drive/MyDrive/DataSet_CourseProject/runs/YOLO26n_baseline/weights/best.pt"
)

# Run validation
results = model.val(
    data="/content/drive/MyDrive/DataSet_CourseProject/PCB_YOLO_Dataset/data.yaml",
    split="val",
    imgsz=640,
    batch=16,
    device=0
)

# Get class names
names = model.names

# Extract class-wise metrics
precision = results.box.p
recall = results.box.r
map50 = results.box.map50
map50_95 = results.box.map

# Create dataframe
df = pd.DataFrame({
    "Class": [names[i] for i in range(len(names))],
    "Precision": precision,
    "Recall": recall,
    "mAP50": map50,
    "mAP50-95": map50_95
})

# Convert to percentages
for col in ["Precision", "Recall", "mAP50", "mAP50-95"]:
    df[col] = (df[col] * 100).round(2)

display(df)
```
